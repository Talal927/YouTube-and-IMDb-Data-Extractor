```python
import requests
from bs4 import BeautifulSoup
import csv
from collections import Counter

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.36'
}

def scrape_and_store_data():
    
    url = 'https://www.imdb.com/chart/top'

    
    response = requests.get(url, headers=headers)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        div_elements = soup.find_all('div', class_='ipc-title ipc-title--base ipc-title--title ipc-title-link-no-icon ipc-title--on-textPrimary sc-4dcdad14-9 dZscOy cli-title')
        movie_data = []

        for div_element in div_elements:
            links = div_element.find_all('a')
            for link in links:
                href = link.get('href')
                movie_url = 'https://www.imdb.com/' + href

                response = requests.get(movie_url, headers=headers)
                if response.status_code == 200:
                    soup = BeautifulSoup(response.text, 'html.parser')
                    title = soup.find(class_="sc-afe43def-1 fDTGTb").text
                    rating = soup.find(class_="sc-bde20123-1 iZlgcd").text

                    
                    release = soup.find('a', href=lambda x: x and "/title/" in x and "/releaseinfo" in x)
                    if release:
                        release_year = release.text
                        if release_year.isnumeric():
                            release_year = int(release_year)
                            if 2013 <= release_year <= 2023:
                                
                                director_container = soup.find('ul', class_='ipc-inline-list ipc-inline-list--show-dividers ipc-inline-list--inline ipc-metadata-list-item__list-content baseAlt')
                                if director_container:
                                    director_link = director_container.find('a', href=lambda x: x and "/name/nm" in x)
                                    if director_link:
                                        director = director_link.text
                                    else:
                                        director = "Director not found"
                                else:
                                    director = "Director container not found"

                                
                                genre_container = soup.find('div', class_='ipc-chip-list--baseAlt ipc-chip-list', attrs={'data-testid': 'genres'})
                                if genre_container:
                                    genres = genre_container.find_all('span', class_='ipc-chip__text')
                                    genre_list = [genre.text for genre in genres]
                                else:
                                    genre_list = ["Genre not found"]

                                movie_data.append([title, rating, release_year, director, ", ".join(genre_list), movie_url])
                else:
                    print("Failed to retrieve movie details for URL:", movie_url)

        
        with open('imdb_top_movies_2013_2023.csv', 'w', newline='', encoding='utf-8') as csvfile:
            csv_writer = csv.writer(csvfile)
            csv_writer.writerow(['Title', 'Rating', 'Release Year', 'Director', 'Genres', 'URL'])
            csv_writer.writerows(movie_data)
    else:
        print("Failed to retrieve the IMDb Top 250 list.")


scrape_and_store_data()

def average_imdb_rating():
    with open('imdb_top_movies_2013_2023.csv', 'r', newline='', encoding='utf-8') as csvfile:
        csv_reader = csv.DictReader(csvfile)
        ratings = [float(row['Rating']) for row in csv_reader]
        average_rating = sum(ratings) / len(ratings)
        return round(average_rating, 1)

def most_common_genre():
    with open('imdb_top_movies_2013_2023.csv', 'r', newline='', encoding='utf-8') as csvfile:
        csv_reader = csv.DictReader(csvfile)
        genres = [row['Genres'].split(', ') for row in csv_reader]
        genre_list = [genre for sublist in genres for genre in sublist]
        most_common = Counter(genre_list).most_common(1)
        return most_common[0][0]

def director_highest_avg_rating():
    with open('imdb_top_movies_2013_2023.csv', 'r', newline='', encoding='utf-8') as csvfile:
        csv_reader = csv.DictReader(csvfile)
        director_ratings = {}
        for row in csv_reader:
            director = row['Director']
            rating = float(row['Rating'])
            if director in director_ratings:
                director_ratings[director].append(rating)
            else:
                director_ratings[director] = [rating]
        highest_avg_director = max(director_ratings, key=lambda k: sum(director_ratings[k]) / len(director_ratings[k]))
        return highest_avg_director

def year_with_most_top_rated_movies():
    with open('imdb_top_movies_2013_2023.csv', 'r', newline='', encoding='utf-8') as csvfile:
        csv_reader = csv.DictReader(csvfile)
        years = [row['Release Year'] for row in csv_reader if row['Release Year'].isnumeric()]
        year_counts = Counter(years)
        most_common_year = max(year_counts, key=year_counts.get)
        return most_common_year


print("Average IMDb Rating for the top-rated movies (2013-2023):", average_imdb_rating())
print("The Most Common Genre among the top-rated movies (2013-2023):", most_common_genre())
print("Director with the Highest Average IMDb Rating (2013-2023):", director_highest_avg_rating())
print("Year with the Highest Number of Top-rated Movies (2013-2023):", year_with_most_top_rated_movies())
```

```python
def discretize_age(age):
    if age <= 35:
        return "Young"
    elif age <= 65:
        return "Middle-aged"
    else:
        return "Old"

df["age_bin"] = df["age"].apply(discretize_age)
```

```python
def remove_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df_filtered = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    return df_filtered

numerical_columns = ['crim', 'zn', 'indus', 'nox', 'rm', 'age', 'dis', 'rad', 'tax', 'ptratio', 'b', 'lstat', 'medv']
for column in numerical_columns:
    df = remove_outliers_iqr(df, column)
```

```python
from scipy import stats
import numpy as np
z_threshold = 3
z_scores = np.abs(stats.zscore(df[numerical_columns]))
outlier_mask = (z_scores > z_threshold).any(axis=1)
df_cleaned = df[~outlier_mask]
df_cleaned.reset_index(drop=True, inplace=True)
```

```python
def normalize(df, column):
    min_value = df[column].min()
    max_value = df[column].max()

    df[column] = (df[column] - min_value) / (max_value - min_value)

    return df

df = normalize(df, "tax")
df = normalize(df, "lstat")
```

```python
from sklearn import linear_model
df=df[["rm","medv"]]
df.head()

num=int(len(df)*0.8)
train=df[:num]
test=df[num:]
print("Data: ",len(df))
print("Train: ",len(train))
print("Test: ",len(test))



def simple_linear_regression(input_feature,output):
    Xi=input_feature
    Yi=output
    
    n=len(Xi)
    
    Xi_mean=Xi.mean()
    
    Yi_mean=Yi.mean()
    
    S_Xi=Xi.sum()
    
    S_Yi=Yi.sum()
    
    S_XiYi=((Xi*Yi).sum())*n
    
    S_Xi_S_Yi=S_Xi*S_Yi
    
    S_XiXi=((Xi*Xi).sum())*n
    
    S_Xi_Square=S_Xi*S_Xi
    
    slope=(S_XiYi-S_Xi_S_Yi)/(S_XiXi-S_Xi_Square)
    
    intercept=Yi_mean-(slope*Xi_mean)
    return slope,intercept


actual_slope,actual_intercept=simple_linear_regression(train["rm"],train["medv"])
print("Slope: ",actual_slope)
print("Intercept: ",actual_intercept)
print("The linear regression model for medv based on rm is: medv = {} + {} * rm".format(actual_intercept, actual_slope))

```
